# UID Analysis: Active/Passive Alternations

**Research questions:**
1. Does active↔passive conversion change the surprisal of the converted sentence?
2. Does genre moderate the effect?
3. Does giving the model more discourse context amplify the CF–factual difference?
4. Does the effect ripple into the following sentence?

**Data files expected in `DATA_DIR`:**
- `sweep_right_word.csv` — word-level UID for s_t, right-only windows (-0,+0) to (-0,+20)
- `next_sent_uid.csv` — word-level UID for s_{t+1} after CF conversion
- `sweep_context_word.csv` — context sweep: s_t scored only, conditioning context varied

```python
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import pingouin as pg
import seaborn as sns
from statsmodels.stats.multitest import multipletests

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 120

DATA_DIR = Path("/Users/nolan/Documents/Data/UID")
```

## 1 — Load & Pair Data

Each CSV contains both factual (`f`) and counterfactual (`cf`) rows. We pair them by
`(doc_name, sent_idx, uid_level, context)` and compute `delta = CF − factual` for each metric.

```python
METRICS = [
    "surp_mean", "surp_slor", "uid_std", "uid_mad", "uid_pwd",
    "uid_range", "uid_slope", "uid_len", "surp_peak", "uid_iqr",
    "uid_skew", "uid_gini", "uid_roughness", "uid_cv",
]

def load_and_pair(path):
    df = pd.read_csv(path, index_col=0, low_memory=False)
    df[["fact", "doc_name", "conv_id", "conv_type"]] = df["doc_id"].str.split("::", expand=True)

    merge_cols = [c for c in ["doc_name", "sent_idx", "uid_level", "context"] if c in df.columns]
    metrics    = [m for m in METRICS if m in df.columns]

    factual = df[df["fact"] == "f"]
    cf      = df[df["fact"] == "cf"]

    paired = cf.merge(factual[merge_cols + metrics], on=merge_cols, suffixes=("_cf", "_f"))
    for m in metrics:
        paired[f"delta_{m}"] = paired[f"{m}_cf"] - paired[f"{m}_f"]

    paired["genre"] = paired["doc_name"].str.split("_").str[1]
    print(f"  {path.name}: {len(cf):,} CF rows → {len(paired):,} paired ({len(paired)/len(cf):.0%})")
    return paired


right = load_and_pair(DATA_DIR / "sweep_right_word.csv")
nxt   = load_and_pair(DATA_DIR / "next_sent_uid.csv")
ctx   = load_and_pair(DATA_DIR / "sweep_context_word.csv") if (DATA_DIR / "sweep_context_word.csv").exists() else None

if ctx is None:
    print("  sweep_context_word.csv not found — run scripts/run_context_sweep.py to generate it")
```

## 2 — Overview

How many pairs do we have per conversion type, and what is the median Δ surp_mean?

```python
base = right[right["uid_level"] == "(-0,+0)"]

print("Conversion type counts (right sweep, no added context):")
print(base["conv_type"].value_counts().to_string())
print()

rows = []
for label, df in [("s_t  (right sweep, -0+0)", base), ("s_t+1 (next sentence)", nxt)]:
    for ct in ["a>p", "p>a"]:
        d = df[df["conv_type"] == ct]["delta_surp_mean"].dropna()
        rows.append({"dataset": label, "conv_type": ct, "n": len(d),
                     "median_Δ": round(d.median(), 4), "mean_Δ": round(d.mean(), 4)})

pd.DataFrame(rows)
```

## 3 — Δ Surprisal Distributions

Distribution of `surp_mean_CF − surp_mean_factual` per sentence pair.
Positive = CF version is more surprising (higher information density).

```python
base = right[right["uid_level"] == "(-0,+0)"]

fig, axes = plt.subplots(1, 2, figsize=(10, 5), sharey=True)
colors = {"a>p": "steelblue", "p>a": "tomato"}

for ax, ct in zip(axes, ["a>p", "p>a"]):
    sub = base[base["conv_type"] == ct]
    sns.violinplot(data=sub, y="delta_surp_mean", inner="box", color=colors[ct], ax=ax)
    ax.axhline(0, color="black", lw=1.2, ls="--")
    ax.set_title(f"{ct}  (n={len(sub):,})")
    ax.set_ylabel("Δ surp_mean [bits/word]" if ct == "a>p" else "")
    ax.set_xlabel("")
    ax.text(0.5, 0.97, f"median = {sub['delta_surp_mean'].median():+.3f}",
            ha="center", va="top", transform=ax.transAxes, fontsize=10)

plt.suptitle("Δ surp_mean: CF − factual  (s_t, right sweep, no added context)", fontsize=13)
plt.tight_layout()
plt.show()
```

## 4 — Genre Breakdown

Median Δ surp_mean by genre. Genres are sorted by number of a>p conversions.

```python
base = right[right["uid_level"] == "(-0,+0)"]

summary = (base.groupby(["genre", "conv_type"])["delta_surp_mean"]
               .agg(median="median", n="count")
               .reset_index())

# Sort genres by a>p count
order = (summary[summary["conv_type"] == "a>p"]
         .sort_values("n", ascending=False)["genre"].tolist())

fig, ax = plt.subplots(figsize=(13, 5))
sns.barplot(data=summary, x="genre", y="median", hue="conv_type",
            order=order, palette={"a>p": "steelblue", "p>a": "tomato"}, ax=ax)

ax.axhline(0, color="black", lw=0.8, ls="--")
ax.set_xlabel("Genre")
ax.set_ylabel("Median Δ surp_mean [bits/word]")
ax.set_title("Conversion effect by genre  (right sweep, -0+0)")
ax.legend(title="conv_type")
plt.tight_layout()
plt.show()

# n per genre
print("\nPairs per (genre, conv_type):")
print(summary.pivot(index="genre", columns="conv_type", values="n").reindex(order).to_string())
```

## 5 — Context Sweep

**Design (collaborator's):** the scored tokens are always s_t only. We vary how much
left context the model is *conditioned on* before scoring.

**Prediction:** if discourse context constrains voice choice, the model should be
increasingly surprised by the "wrong" voice as context grows → delta grows with context.

**Contrast with the windowed sweep:** in the window sweep the left tokens were *scored*,
not just conditioned on — those tokens are identical between CF and factual, so they dilute
the measured delta. This sweep avoids that confound entirely.

```python
if ctx is None:
    print("Context sweep data not available. Run scripts/run_context_sweep.py first.")
else:
    CTX_TOKENS = {
        "no_ctx":        0,
        "tok[-10,+0]":  10,
        "tok[-20,+0]":  20,
        "tok[-50,+0]":  50,
        "tok[-100,+0]": 100,
        "tok[-200,+0]": 200,
        "document":     300,   # placeholder x-position; labelled on axis
    }

    summary = (ctx.groupby(["context", "conv_type"])["delta_surp_mean"]
                  .agg(median="median",
                       sem=lambda x: x.std() / np.sqrt(len(x)),
                       n="count")
                  .reset_index())
    summary["ctx_x"] = summary["context"].map(CTX_TOKENS)
    summary = summary.dropna(subset=["ctx_x"]).sort_values("ctx_x")

    fig, ax = plt.subplots(figsize=(10, 5))
    for ct, color in [("a>p", "steelblue"), ("p>a", "tomato")]:
        sub = summary[summary["conv_type"] == ct]
        ax.plot(sub["ctx_x"], sub["median"], marker="o", label=ct, color=color)
        ax.fill_between(sub["ctx_x"],
                        sub["median"] - 1.96 * sub["sem"],
                        sub["median"] + 1.96 * sub["sem"],
                        alpha=0.15, color=color)
        for _, row in sub.iterrows():
            ax.annotate(f"n={row['n']:.0f}",
                        (row["ctx_x"], row["median"]),
                        textcoords="offset points", xytext=(0, 9), ha="center", fontsize=7)

    ax.axhline(0, color="black", lw=0.8, ls="--")
    ax.set_xlabel("Conditioning context (left tokens given to model)")
    ax.set_ylabel("Median Δ surp_mean [bits/word]")
    ax.set_title("Context sweep: does more discourse context amplify the CF–factual difference?\n"
                 "(scored tokens = s_t only)")
    ax.legend(title="conv_type")

    xtick_items = sorted(CTX_TOKENS.items(), key=lambda x: x[1])
    ax.set_xticks([v for _, v in xtick_items])
    ax.set_xticklabels([k for k, _ in xtick_items], rotation=30, ha="right")

    plt.tight_layout()
    plt.show()
```

## 6 — Next Sentence Effect

Does converting s_t change how surprising the following sentence s_{t+1} is?
The text of s_{t+1} is identical between conditions — any delta is a pure discourse context effect.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5), sharey=True)

for ax, ct in zip(axes, ["a>p", "p>a"]):
    sub = nxt[nxt["conv_type"] == ct]
    sns.violinplot(data=sub, y="delta_surp_mean", inner="box",
                   color="steelblue" if ct == "a>p" else "tomato", ax=ax)
    ax.axhline(0, color="black", lw=1.2, ls="--")
    ax.set_title(f"{ct}  (n={len(sub):,})")
    ax.set_ylabel("Δ surp_mean at s_t+1 [bits/word]" if ct == "a>p" else "")
    ax.set_xlabel("")
    ax.text(0.5, 0.97, f"median = {sub['delta_surp_mean'].median():+.4f}",
            ha="center", va="top", transform=ax.transAxes, fontsize=10)

plt.suptitle("Δ surp_mean for s_t+1  (next sentence after conversion)", fontsize=13)
plt.tight_layout()
plt.show()
```

## 7 — Statistical Tests (Wilcoxon Signed-Rank)

Non-parametric paired test with Bonferroni correction across all windows and conversion types.

```python
def wilcoxon_table(paired_df, label, metric="surp_mean"):
    rows = []
    levels = paired_df["uid_level"].unique() if "uid_level" in paired_df.columns else ["sentence"]
    for lvl in levels:
        sub_df = paired_df[paired_df["uid_level"] == lvl] if "uid_level" in paired_df.columns else paired_df
        for ct in ["a>p", "p>a"]:
            sub = sub_df[sub_df["conv_type"] == ct][[f"{metric}_cf", f"{metric}_f"]].dropna()
            if len(sub) < 10:
                continue
            res = pg.wilcoxon(sub[f"{metric}_cf"], sub[f"{metric}_f"])
            delta = sub[f"{metric}_cf"] - sub[f"{metric}_f"]
            rows.append({
                "dataset":   label,
                "window":    lvl,
                "conv_type": ct,
                "n":         len(sub),
                "median_Δ":  round(delta.median(), 4),
                "W":         res["W_val"].iloc[0],
                "p":         res["p_val"].iloc[0],
                "RBC":       round(res["RBC"].iloc[0], 3),
            })
    return pd.DataFrame(rows)


results = pd.concat([
    wilcoxon_table(right, "right sweep"),
    wilcoxon_table(nxt,   "next sentence"),
], ignore_index=True)

_, p_bonf, _, _ = multipletests(results["p"], method="bonferroni")
results["p_bonf"] = p_bonf.round(4)
results["sig"]    = results["p_bonf"] < 0.05

print("RBC: |r| < 0.1 negligible, 0.1–0.3 small, 0.3–0.5 medium, >0.5 large")
results.sort_values(["dataset", "window", "conv_type"])
```
